```python
import pandas as pd
df = pd.read_csv('../data/students.csv')
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['math score'].mean()
```

```python
import matplotlib.pyplot as plt

plt.bar(['Math','Reading','Writing'],
        [df['math score'].mean(),
         df['reading score'].mean(),
         df['writing score'].mean()])
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

X = df[['reading score']]
y = df['writing score']

model = LinearRegression()
model.fit(X, y)

new_data = pd.DataFrame({'reading score': [70]})
prediction = model.predict(new_data)

print("Predicted Writing Score:", prediction[0])
```
